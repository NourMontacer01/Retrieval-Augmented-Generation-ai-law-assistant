# Démonstration : recherche et réponse citée

À ouvrir depuis le dossier `notebooks/`, après la construction de la base (`lancer.bat`, choix 2).

```python
import sys
sys.path.insert(0, "..")  # la racine du projet, qui contient le paquet conformite

from conformite.recherche import Recherche, afficher
from conformite.reponse import repondre

recherche = Recherche()  # charge le modele et la base une seule fois
```

## 1. Recherche seule

Les passages les plus proches de la question, avec leur nature juridique et leur score de proximité (1 = identique).

```python
passages = recherche.rechercher("Dans quel delai notifier une violation de donnees ?", k=3)
afficher(passages)
```

## 2. Filtrer sur les étiquettes

Le filtre s'applique avant le classement : on ne cherche que dans l'AI Act, et seulement dans les articles.

```python
passages = recherche.rechercher("Qu'est-ce qu'un systeme d'IA a haut risque ?", k=3, source_id="ai-act", type_unite="article")
afficher(passages)
```

## 3. Réponse rédigée

Nécessite une clé `MISTRAL_API_KEY` dans `.env`. Sans clé, seuls les passages sont renvoyés.

```python
resultat = repondre(recherche, "Quelles pratiques d'IA sont interdites ?")
print(resultat["reponse"] or "(pas de cle d'API)")
for i, p in enumerate(resultat["passages"], 1):
    print(f"[{i}] {p['reference']} - {p['nature']}")
```
